### Exploratory Data Analysis


**1**:Compute total fee revenue (transaction_fee_kes) per calendar month, then add a column showing the month-over-month % change. Which month grew fastest? (Skills: time bucketing, LAG() / .pct_change().)

```python
#Create  new dataframe with only completed transactions
completed = data[data['status']=='Completed']
```

```python
# Group by month and sum fees
monthly_revenue = (
                   completed
                   .set_index("transaction_timestamp")
                   .resample("ME")["transaction_fee_kes"]
                   .sum()
                   .reset_index()
)

print(monthly_revenue)
```

```python
# Visualizing monthly revenue overtime
plt.figure(figsize=(7,5))
sns.barplot(
    data=monthly_revenue,
    x="transaction_timestamp",
    y="transaction_fee_kes")
plt.title("Monthly Revenue")
plt.xlabel("transaction_timestamp")
plt.ylabel("transaction_fee_kes")
plt.xticks(rotation=45)
plt.show()
```

```python
## Month over Month percentage  change
monthly_revenue['MOM_growth_pct'] = (
    monthly_revenue["transaction_fee_kes"]
    .pct_change() * 100)
monthly_revenue
```

```python
plt.figure(figsize=(7,5))
sns.lineplot(
    monthly_revenue,
    x="transaction_timestamp",
    y="MOM_growth_pct")
plt.axhline(
    y=0,
    color="red",
    linestyle="--")
plt.title("Monthly Revenue Growth")
plt.xlabel("Months")
plt.ylabel("Pct Growth")
plt.xticks(rotation=45)
plt.show()
```

__Conclusion__
- The month with the highest month over month positive percentage change was June 

- July had the highest negative month over month percentage

**2.** Peak transaction hours (heatmap-ready)

Build a day-of-week × hour-of-day matrix of transaction counts. Identify the single busiest hour block. (Skills: extracting hour & weekday from a timestamp, pivoting.)

**Pivot Table**

```python
completed['weekday'] = completed['transaction_timestamp'].dt.day_name()
completed['hour'] = completed['transaction_timestamp'].dt.hour
completed.head()
```

```python
pivot_table = completed.pivot_table(
    index = 'hour',
    columns = 'weekday',
    values = 'transaction_id',
    aggfunc = 'count',
    fill_value = 0
)

print(pivot_table)
```

```python
peak = (
    completed
    .groupby(['weekday',
              'hour'])
    .size()
    .reset_index(name = 'transactions')
    .sort_values('transactions',
                 ascending = False)
                 )

print(peak)
```

_Conclusion_
- The highest number of transactions we're conducted on Wednesday at 9 am

**3.** Top customers by value, ranked within their county

**4.** 7-day rolling transaction volume

Create a daily transaction-count series, then compute the 7-day rolling average of daily volume. On how many days did volume exceed 1.5× its rolling average? (Skills: resampling to daily, rolling windows.)

_Info needed_
- transaction_id
- transaction_timestamp
- day

```python
daily = completed.set_index('transaction_timestamp').resample('D')['transaction_id'].count()

print(daily)
```

```python
rolling = daily.rolling(7).mean()

results = pd.DataFrame({
    'daily': daily,
    'rolling_average': rolling
})

results['Above_1.5_times'] = results['daily'] > 1.5 * results ['rolling_average']

print(results)
```

**5.** Customer RFM segmentation

For every customer compute Recency (days since last transaction, relative to the dataset's max date), Frequency (number of transactions), and Monetary (total amount). Score each into quintiles (1–5) and label the top segment as "Champions". (Skills: multi-metric aggregation, quantile bucketing, NTILE().)

_Info needed_
- transaction_timestamp
- transaction_id
- amount_kes
- customer_id

```python
latest = completed['transaction_timestamp'].max()
rfm = completed.groupby('customer_id').agg(
    Recency = (
        'transaction_timestamp',
        lambda x:(latest - x
        .max())
        .days),
    Frequency = (
        'transaction_id',
        'count'),
    Monetary = (
        'amount_kes',
        'sum')
)

rfm['R'] = pd.qcut(
    rfm['Recency'],
    5,
    labels = [5, 4, 3, 2, 1])

rfm['F'] = pd.qcut(
    rfm['Frequency']
    .rank(method = 'first'),
    5,
    labels = [1, 2, 3, 4, 5])

rfm['M'] = pd.qcut(
    rfm['Monetary'],
    5,
    labels = [1, 2, 3, 4, 5])

rfm['segment'] = np.where(
    (rfm['R'] == 5) &
    (rfm['F'] == 5) &
    (rfm['M'] == 5),
    'Champions',
    'Other'
) 

rfm.head()
```
